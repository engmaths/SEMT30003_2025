## Imports

```python
from pylab           import * 
from hhhelper        import *  
from scipy.integrate import solve_ivp 
from seaborn         import despine

# Plotting stuff
rc('lines',lw=0.8) # change all line widths
def quickaxis(xl="",yl="",mytitle=""):
    # Tidy up and label an axis
    xlabel(xl); ylabel(yl); title(mytitle);
    autoscale(enable=True, axis='x', tight=True)
    despine(bottom=True)
```

## Simulation code

Write code to visualize the model states once.
Change the parameters to `HH_model()` to and re-run this cell to answer individual questions. 
(Or, copy and past this cell for each question, then change the model parameters, if you'd like a record of the output). 

```python
I  = 0.7 # nA
T  = 200 # ms
tt = linspace(0,T,T*10+1)
HH = HH_model()

v,m,n,h   = solve_ivp(lambda t,state:HH.ode(*state,I),(0,T),[-70,0,0,0],"RK45",tt).y
IL,IK,INa = HH.currents(v,m,n,h)

# Plotting stuff
figure(0,(6,5),90)
subplot(311); 
plot(tt,v); 
quickaxis('','$v(t)$ (mV)',"Membrane voltage")
subplot(312); 
plot(tt,n,label='K$^+$  n gate')
plot(tt,m,label='Na$^+$ m gate')
plot(tt,h,label='Na$^+$ h gate')
quickaxis('','Fraction\nOpen',"Gating Variables")
legend(bbox_to_anchor=(1,1),fontsize=8,frameon=False)
subplot(313); 
plot(tt,IL,label='Leak current')
plot(tt,IK,label='Potassium current')
plot(tt,INa,label='Sodium current')
quickaxis("Time (ms)",'Current (nA)',"Currents (nA)")
legend(bbox_to_anchor=(1,1),fontsize=8,frameon=False)
tight_layout()
```
